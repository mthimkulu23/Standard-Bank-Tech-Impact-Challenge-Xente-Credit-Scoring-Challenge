```python
import pandas as pd

#  Load data
Train = pd.read_csv('Train.csv')
Train
```

```python
Trains = Train.dropna()
Trains.isnull().sum()
```

```python
remove_dupli_tr = Trains.drop_duplicates()
remove_dupli_tr
```

```python
y = remove_dupli_tr['IsDefaulted'].astype(int) #DISPLAYING Y
y[y == 1]
y[y == 0]
```

```python
x = remove_dupli_tr.drop(["AmountLoan","Currency","PaidOnDate","IsFinalPayBack","DueDate","PayBackId","IsThirdPartyConfirmed","IsDefaulted"], axis =1 )
x
```

```python
#Load data
test = pd.read_csv('Test.csv')
test
```

```python
tests = test.dropna()
tests


xt = tests.drop_duplicates()
xt
```

```python
# convert to datetime train_new

x['IssuedDateLoan'] = pd.to_datetime(x['IssuedDateLoan'], errors='coerce')
x['TransactionStartTime'] = pd.to_datetime(x['TransactionStartTime'], errors='coerce')

# # Loan issue date features
x['issue_loan_year'] = x['IssuedDateLoan'].dt.year
x['issue_loan_month'] = x['IssuedDateLoan'].dt.month
x['issue_loan_day'] = x['IssuedDateLoan'].dt.day
x['issue_loan_time'] = x['IssuedDateLoan'].dt.time
x['issue_loan_hour'] = x['IssuedDateLoan'].dt.hour

# # # Transaction time features
x['hour'] = x['TransactionStartTime'].dt.hour
x['trans_time'] = x['TransactionStartTime'].dt.time
x['trans_year'] = x['TransactionStartTime'].dt.year
x['trans_month'] = x['TransactionStartTime'].dt.month
x['trans_day'] = x['TransactionStartTime'].dt.day


x.drop(['IssuedDateLoan','TransactionStartTime'], axis=1, inplace=True)

x







```
